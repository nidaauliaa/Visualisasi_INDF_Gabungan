```python
import pandas as pd

df = pd.read_csv("gabungan.csv")

currency_cols = [col for col in df.columns if col not in ["Hari Ke-", "Tanggal"]]
df[currency_cols] = df[currency_cols].astype(float).round(0).astype(int)

with pd.ExcelWriter("gabungan_currency.xlsx", engine="xlsx") as writer:
    df.to_excel(writer, index=False, sheet_name="Sheet1")
    workbook = writer.book
    worksheet = writer.sheets["Sheet1"]
    currency_format = workbook.add_format({"num_format": 'Rp#,##0'})
    worksheet.set_column(2, len(df.columns) - 1, 15, currency_format)
```
